## Concepts

**Agentic customization** means picking the right extension mechanism:

- **Built-in tools**: Anthropic-provided capabilities (e.g., web/code tools where available). Fast to adopt; limited to what Anthropic ships and cannot reach arbitrary internal systems.
- **Custom tools**: your own functions via the tool-use API. Full control; app-specific and lives inside one application.
- **Skills**: packaged, reusable instructions/capabilities (especially in Claude Code) that guide how Claude performs a class of tasks.
- **MCP servers**: reusable tools/resources/prompts shared across many apps and maintained independently - best when multiple applications need the same capability.

Rule of thumb: one app + bespoke logic -> custom tool; many apps + shared capability -> MCP; reusable task know-how -> Skill; covered by a built-in -> use the built-in.

```python
def recommend(scenario):
    if scenario.get("covered_by_builtin"):
        return "built-in tool"
    if scenario.get("shared_across_apps") and scenario.get("needs_live_system"):
        return "MCP server"
    if scenario.get("reusable_task_knowledge"):
        return "Skill"
    return "custom tool"

# Sample Q3: internal REST API reused across several apps, maintained independently
print(recommend({"shared_across_apps": True, "needs_live_system": True}))
print(recommend({"reusable_task_knowledge": True}))
print(recommend({}))
```
