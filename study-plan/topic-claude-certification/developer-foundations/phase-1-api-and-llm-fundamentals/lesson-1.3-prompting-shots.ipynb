## Concepts

The number of worked **examples** you give shapes reliability:

- **Zero-shot**: instructions only, no examples. Fast and cheap; best when the task is common and the output format is simple.
- **Single-shot**: one example to anchor format/tone.
- **Multi-shot (few-shot)**: several examples covering edge cases - best for strict formats, classification, and unusual styles.

In the Messages API you can embed examples as alternating `user`/`assistant` turns (the model treats prior assistant turns as demonstrations) or inline them in a single user message. More examples cost more tokens, so stop once accuracy plateaus.

## Code-Along: Step 1 — Zero-shot classification

Ask for a label with instructions only.

```python
def classify(messages_payload):
    if not HAVE_API:
        return "[MOCK label]"
    msg = client.messages.create(
        model=MODEL_HAIKU, max_tokens=10, messages=messages_payload,
    )
    return msg.content[0].text.strip()

zero = [{"role": "user", "content": "Classify sentiment as positive/negative: 'I love this!'"}]
print("zero-shot:", classify(zero))
```

## Code-Along: Step 2 — Few-shot with example turns

Demonstrate the exact label format using prior assistant turns.

```python
few = [
    {"role": "user", "content": "Sentiment: 'Worst purchase ever.'"},
    {"role": "assistant", "content": "negative"},
    {"role": "user", "content": "Sentiment: 'Absolutely delightful.'"},
    {"role": "assistant", "content": "positive"},
    {"role": "user", "content": "Sentiment: 'It arrived on time.'"},
]
print("few-shot:", classify(few))
```
