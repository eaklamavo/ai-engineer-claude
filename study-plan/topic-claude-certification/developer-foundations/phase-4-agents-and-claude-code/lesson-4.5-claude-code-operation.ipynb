## Concepts

**Claude Code** is Anthropic's agentic coding tool. Core components:

- **Rules/Memory (CLAUDE.md)**: persistent project/user instructions. The **CLAUDE.md hierarchy** layers enterprise -> user (`~/.claude/CLAUDE.md`) -> project (`./CLAUDE.md`) -> subdirectory files, merged so more specific files refine broader ones.
- **settings.json**: configuration (permissions, model, tools, env) at user/project levels.
- **Skills**: packaged task know-how; **Commands**: custom **slash commands** (Markdown files in `.claude/commands/`); **Agents**: specialized subagents.
- **Modes**: interactive, **headless** (non-interactive, scriptable via `claude -p`), **streaming** output, and **auto** mode for autonomous multi-step work.

> Flags and file locations evolve; confirm current Claude Code settings at https://docs.anthropic.com/en/docs/claude-code.

> **Reference:** [Anthropic - Claude Code](https://docs.anthropic.com/en/docs/claude-code/overview)

## Code-Along: Step 1 — Author a project CLAUDE.md

Project memory/rules that Claude Code loads automatically.

```python
%%writefile CLAUDE.md
# Project: Inventory Agent

## Rules
- Use Python 3; follow PEP 8.
- Never call mutating tools without an approval hook.
- Prefer the `anthropic` SDK; pin model versions.

## Context
- The MCP inventory server lives in `inventory_server.py`.
- Cost target: prefer Haiku for extraction, Sonnet for synthesis.
```

## Code-Along: Step 2 — Author a settings.json

Project-level configuration for Claude Code.

```python
%%writefile settings.json
{
  "model": "claude-sonnet-4-5",
  "permissions": {
    "allow": ["Read", "Edit"],
    "ask": ["Bash"]
  },
  "env": {
    "ANTHROPIC_API_KEY": "${ANTHROPIC_API_KEY}"
  }
}
```

## Code-Along: Step 3 — Create a custom slash command

A Markdown file in `.claude/commands/` becomes `/review`.

```python
import os
os.makedirs(".claude/commands", exist_ok=True)
with open(".claude/commands/review.md", "w", encoding="utf-8") as f:
    f.write("Review the staged diff for bugs, security issues, and missing tests. "
            "Report findings as a bulleted list grouped by severity.\n")
print("created .claude/commands/review.md -> use as /review")
```
