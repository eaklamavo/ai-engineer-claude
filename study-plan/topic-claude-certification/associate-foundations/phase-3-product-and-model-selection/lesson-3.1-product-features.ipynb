## Concepts

Claude offers several product surfaces; choosing well saves time and money.

- **Chat** — fast, conversational, one-off tasks.
- **Projects** — a reusable workspace with saved **instructions** and **knowledge sources**; ideal for recurring workflows where consistent context matters.
- **Research mode** — for deeper, multi-step research that gathers and synthesizes information across sources. (Verify current availability/name at docs.anthropic.com - features evolve.)
- **Artifacts** — a standalone, editable deliverable panel (documents, tables, drafts, code).

Rule of thumb: *one-off -> chat; repeatable -> Project; deep multi-source research -> research mode; editable deliverable -> artifact.*

This maps to Domain 3: *Select appropriate Claude product features*.

> **Reference:** [Anthropic Docs - Claude.ai features (Projects)](https://support.anthropic.com)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Map needs to features

Run to see the mapping you'll memorize.

```python
features = {
 'Quick one-off question': 'Chat',
 'Recurring workflow with the same context each time': 'Project',
 'Deep, multi-source investigation': 'Research mode',
 'An editable document/table deliverable': 'Artifact',
}
for need, feat in features.items():
    print(f'- {need} -> {feat}')
```
