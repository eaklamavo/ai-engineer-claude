## Concepts

Claude reads a limited amount of text at once - its **context window**. A very long chat can push earlier details out of view, so Claude may 'forget' things said much earlier or get less sharp.

Signs to act:
- Claude repeats itself, loses the thread, or contradicts earlier points.
- The conversation has drifted across many topics.

Three moves:
- **Restart** — begin a fresh chat for a new topic.
- **Summarize** — ask Claude to summarize the key decisions so far, then continue from that summary.
- **Persist** — put durable context in a **Project** (instructions + knowledge) so it's always available, instead of re-pasting it each time.

This maps to Domain 3: *Understand and manage context limitations and memory (when to restart, summarize, or persist)*.

> **Reference:** [Anthropic Docs - Context windows](https://docs.anthropic.com/en/docs/build-with-claude/context-windows)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Summarize-to-continue in action

Compress a long thread into a carry-forward summary.

```python
long_thread = 'We decided: launch date = May 1; budget = $5k; owner = Priya; channel = email + LinkedIn; metric = signups.'
summary = ask_claude(f'Summarize the key decisions below as a short bullet list I can paste into a new chat to continue.\n\n{long_thread}')
print(summary)
```

## Code-Along: Step 2 — Restart vs. summarize vs. persist

A quick decision guide.

```python
guide = {
 'Brand-new, unrelated topic': 'Restart (new chat)',
 'Same topic but the chat is long/confused': 'Summarize, then continue',
 'Context you reuse every time': 'Persist in a Project',
}
for sit, move in guide.items():
    print(f'- {sit} -> {move}')
```
