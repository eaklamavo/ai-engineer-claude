## Concepts

Good model selection is a **trade-off decision**. Ask:

- **How complex is the reasoning?** Simple classification/short drafts -> smaller model. Nuanced analysis -> larger model.
- **How much volume / how fast?** High volume or real-time -> prioritize speed and cost (smaller model).
- **How high are the stakes?** High-impact, complex, or sensitive output -> invest in the more capable model and human review.

Sample Q2's lesson: match a *faster, lower-cost* model to straightforward, high-volume work and reserve the top model for complex reasoning. The goal is the **right** model, not the biggest.

This maps to Domain 3: *Align model selection with task requirements (cost, speed, quality)*.

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — A simple model-picker

Answer three questions, get a suggestion.

```python
def pick_model(complex_reasoning, high_volume, high_stakes):
    if complex_reasoning or high_stakes:
        return 'Opus-class (most capable) + human review'
    if high_volume:
        return 'Haiku-class (fast, low cost)'
    return 'Sonnet-class (balanced)'

print(pick_model(complex_reasoning=False, high_volume=True, high_stakes=False), '<- many short replies')
print(pick_model(complex_reasoning=True, high_volume=False, high_stakes=True), '<- complex strategy memo')
print(pick_model(complex_reasoning=False, high_volume=False, high_stakes=False), '<- everyday drafting')
```
