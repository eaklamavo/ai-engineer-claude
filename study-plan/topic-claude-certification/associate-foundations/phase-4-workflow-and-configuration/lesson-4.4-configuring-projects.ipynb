## Concepts

A **Project** gives Claude consistent context every time via two things:

- **Instructions (system-level)** — persistent guidance: the role, tone, rules, do's/don'ts, and output format. Think of it as onboarding Claude once instead of re-explaining each chat. Good instructions are specific, scoped, and include guardrails (e.g., 'always flag claims that need verification').
- **Knowledge sources** — files/documents you upload (or connect) so Claude can ground answers in your real material (policies, brand guide, FAQs).

Together they make outputs consistent and relevant. **Before uploading, apply data-sensitivity rules** (covered in Phase 5): anonymize regulated personal data first.

This maps to Domain 5: *Configure Claude Projects with instructions and knowledge sources* and *create effective system-level instructions*.

> **Reference:** [Anthropic Docs - Projects (Claude.ai help)](https://support.anthropic.com)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Draft strong project instructions

Ask Claude to help write reusable instructions.

```python
instr = ask_claude('Write 6 concise project instructions for an assistant that drafts customer support replies in a friendly, accurate tone and flags anything needing human review. Bullet list.')
print(instr)
```
