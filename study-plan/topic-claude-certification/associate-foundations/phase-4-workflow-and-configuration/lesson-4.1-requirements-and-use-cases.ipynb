## Concepts

Before automating anything, understand the **requirements** and find the best **use cases**.

- **Analyze requirements** — what is the goal, who are the users, what inputs/outputs, what constraints? Claude can help you draft and question a requirements list.
- **Spot use cases** — look for tasks that are repetitive, text-heavy, or time-consuming: summarizing, drafting, categorizing, first-pass analysis.
- **Prioritize** — rank by **value** (time saved, quality gained) vs. **effort/risk**. Start with high-value, low-risk tasks.

Associates identify opportunities and escalate complex technical builds to Architects/Developers.

This maps to Domain 4: *Apply Claude to analyze requirements and use cases*.

> **Reference:** [Anthropic Docs - Use case guides](https://docs.anthropic.com/en/docs/about-claude/use-case-guides/overview)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Draft requirements with Claude

Claude as a thinking partner for scoping.

```python
reqs = ask_claude('I want to speed up our weekly newsletter. Ask me 4 clarifying questions, then list likely requirements. Keep it short.')
print(reqs)
```

## Code-Along: Step 2 — Surface candidate use cases

Brainstorm where Claude could help in a workflow.

```python
cases = ask_claude('For a small marketing team, list 6 repetitive text tasks where Claude could save time. One line each.')
print(cases)
```
