## Concepts

Associates often need to **explain Claude to others** - a manager, a team, a client. Good communication covers both sides:

- **Value** — time saved, faster drafts, consistent quality, capacity for higher-value work.
- **Limitations** — can hallucinate, may be out of date, needs human review for high-stakes work, isn't a substitute for expert judgment, and data-sensitivity rules apply (Phase 5).

Honest framing builds trust and sets realistic expectations - over-promising leads to misuse and disappointment.

This maps to Domain 4: *Communicate Claude's value and limitations to stakeholders*.

> **Reference:** [Anthropic Docs - Overview](https://docs.anthropic.com)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Draft a balanced briefing

Ask for value AND limitations together.

```python
brief = ask_claude('Write a 6-line briefing for a manager: 3 lines on how Claude helps our team and 3 lines on its limitations/where human review is needed. Plain language.')
print(brief)
```
