## Concepts

Integration means fitting Claude into a **real process**, not using it in isolation. A good approach:

1. **Map the current steps** (as-is).
2. **Identify where Claude adds value** (draft, summarize, analyze first pass).
3. **Insert human checkpoints** - especially review/validation steps (from Phase 2) for anything high-stakes.
4. **Define the new flow** (to-be) and who owns each step.

Claude usually **augments** a workflow (speeds a step) rather than replacing the human judgment around it.

This maps to Domain 4: *Integrate Claude into existing workflows to augment or redesign them* and *leverage Claude for research, planning, and process optimization*.

> **Reference:** [Anthropic Docs - Use case guides](https://docs.anthropic.com/en/docs/about-claude/use-case-guides/overview)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Draft a to-be workflow

Ask Claude to insert itself sensibly, with review points.

```python
flow = ask_claude('Here is a process: collect feedback -> write summary -> send to manager. Rewrite it showing where Claude helps and where a human must review. Short numbered steps.')
print(flow)
```

## Exercise

Map one of your workflows before and after adding Claude. Fill in the TODO steps and mark review points, then save.

```python
as_is = ['TODO step 1','TODO step 2','TODO step 3']
to_be = ['TODO step 1 (Claude drafts)','TODO step 2 (human review)','TODO step 3 (send)']
lines = ['# Workflow map','', '## As-is'] + [f'{i}. {s}' for i,s in enumerate(as_is,1)]
lines += ['','## To-be (with Claude + human checkpoints)'] + [f'{i}. {s}' for i,s in enumerate(to_be,1)]
with open('workflow_map.md','w',encoding='utf-8') as f:
    f.write('\n'.join(lines))
print('Saved workflow_map.md')
print('\n'.join(lines))
```

## Key Takeaways

- Map the current process before inserting Claude
- Add human checkpoints for high-stakes steps
- Claude augments workflows; humans keep judgment and ownership
- Define who owns each step in the new flow
