## Concepts

When an output disappoints, troubleshoot systematically instead of guessing. Common causes and fixes:

- **Too vague** -> add role, context, and format (Phase 1).
- **Missing parts** -> decompose into steps (Phase 1.3).
- **Wrong facts** -> provide the source; fact-check; verify (Phase 2).
- **Wrong tone/length** -> specify audience and constraints; iterate (Phase 1.4).
- **Lost the thread** -> summarize or restart; persist context in a Project (Phase 3.4).
- **Wrong tool/model** -> switch feature or model (Phase 3).

Adjust based on **feedback and results**: change one thing at a time so you learn what worked.

This maps to Domain 7: *Identify, diagnose, and resolve issues with underperforming prompts or poor outputs* and *adjust approach based on feedback*.

> **Reference:** [Anthropic Docs - Prompt engineering overview](https://docs.anthropic.com/en/docs/build-with-claude/prompt-engineering/overview)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Symptom -> fix lookup

A quick diagnostic table.

```python
fixes = {
 'Output is generic/vague': 'Add role, context, and explicit format',
 'Missed part of the request': 'Decompose into ordered steps',
 'Contains wrong facts': 'Provide the source; verify; fact-check',
 'Wrong tone or too long': 'Specify audience + constraints; iterate',
 'Claude lost track': 'Summarize or restart; persist context in a Project',
 'Too slow/expensive or too weak': 'Switch model/feature',
}
for symptom, fix in fixes.items():
    print(f'- {symptom}\n    -> {fix}')
```

## Code-Along: Step 2 — Fix a weak prompt live

Improve a vague prompt and compare.

```python
weak = 'Write something about our quarterly results.'
improved = ('Role: finance communications writer.\n'
            'Task: Summarize Q3 results for staff.\n'
            'Context: Revenue up 8%, costs flat, one new product launched.\n'
            'Format: 4 bullets, plain language, no jargon.')
print('WEAK ->', ask_claude(weak))
print('\nIMPROVED ->', ask_claude(improved))
```
