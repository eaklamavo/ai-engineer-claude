## Concepts

Optimization makes a working process **faster, cheaper, or better** without losing quality.

Techniques:
- **Reuse** — turn repeated prompts into **templates** and recurring context into **Projects** (Phase 4).
- **Right-size** — match the model/feature to the task (Phase 3) to cut cost/latency.
- **Reduce rework** — add review checkpoints early so errors are caught before they spread (Phase 2).
- **Standardize** — consistent instructions reduce variability.
- **Measure** — track a simple before/after metric (time per task, number of revisions, error rate). If it didn't improve a metric, it isn't an optimization.

This maps to Domain 7: *Optimize workflows for efficiency and effectiveness*.

> **Reference:** [Anthropic Docs - Use case guides](https://docs.anthropic.com/en/docs/about-claude/use-case-guides/overview)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Find the bottleneck

Ask Claude to critique a workflow for waste.

```python
wf = 'Each week I re-type the same prompt, paste 3 files, draft a report, then fix tone twice.'
ideas = ask_claude(f'Suggest 4 ways to make this weekly workflow faster using templates, Projects, and model choice. One line each.\n\n{wf}')
print(ideas)
```

## Code-Along: Step 2 — Measure before/after

A tiny metric tracker.

```python
def improvement(before, after):
    saved = before - after
    pct = (saved / before * 100) if before else 0
    return f'Saved {saved} min/task ({pct:.0f}% faster)'
print(improvement(before=30, after=12))
```

## Exercise

Write an optimization plan with real numbers. Fill the TODOs and save.

```python
before_min = 30  # TODO your current time per task
after_min = 12   # TODO expected/new time per task
changes = ['Create a reusable prompt template','Move recurring context into a Project','Right-size the model','TODO your own change']
lines = ['# Workflow optimization plan','', '## Changes'] + [f'- {c}' for c in changes]
lines += ['','## Metric', improvement(before_min, after_min)]
with open('optimization_plan.md','w',encoding='utf-8') as f:
    f.write('\n'.join(lines))
print('Saved optimization_plan.md')
print('\n'.join(lines))
```
