## Concepts

A correct output still needs to fit its **audience** and arrive in the right **format**.

- **Adapt for audience** — a message for executives (brief, outcomes) differs from one for a technical team (detail) or customers (plain, friendly). Edit tone, length, and jargon accordingly.
- **Compare and refine** — generate alternatives and pick/merge the best, rather than accepting the first.
- **Choose the format:**
  - **Inline** chat reply — quick answers and short content.
  - **Artifact** — a standalone, editable deliverable (document, table, long draft) you'll revise.
  - **Structured data** (table, list, CSV-like) — when the result feeds a process or comparison.

This maps to Domain 2: *Edit, adapt, refine, and compare outputs* and *select appropriate output formats*.

> **Reference:** [Anthropic Docs - Artifacts (Claude.ai help)](https://support.anthropic.com)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Same content, two audiences

Adapt one message for two readers.

```python
base = 'We are moving the team meeting from Monday to Wednesday starting next week.'
exec_v = ask_claude(f'Rewrite for a busy executive: one sentence, outcome-focused.\n\n{base}')
team_v = ask_claude(f'Rewrite as a friendly team note with the reason and the new time blank to fill in.\n\n{base}')
print('EXEC:\n', exec_v, '\n\nTEAM:\n', team_v)
```

## Code-Along: Step 2 — Pick the right format

A quick rule you can reuse.

```python
guide = {
 'Short answer or quick reply': 'Inline chat',
 'A document/draft you will edit and reuse': 'Artifact',
 'A comparison or data for a process': 'Structured data (table/list)',
}
for need, fmt in guide.items():
    print(f'- {need} -> {fmt}')
```

## Key Takeaways

- Tailor tone, length, and jargon to the specific audience
- Generate and compare alternatives instead of accepting the first draft
- Use artifacts for editable deliverables, inline for quick replies
