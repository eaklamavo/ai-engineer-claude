## Concepts

Two different questions matter when you review an output:

- **Accuracy** — Is it *correct*? Are the facts, numbers, names, and citations true?
- **Completeness** — Does it cover *everything you asked for*? Any missing section, step, or requirement?

A fluent, confident-sounding answer can still be inaccurate or incomplete. Reviewing is your job, not Claude's. A simple evaluation checklist:
1. Did it answer the actual question?
2. Are all requested parts present?
3. Are factual claims verifiable?
4. Any numbers/names/dates/citations to double-check?
5. Is anything contradicting itself?

This maps to Domain 2: *Evaluate Claude-generated outputs for accuracy and completeness*.

> **Reference:** [Anthropic Docs - Reduce hallucinations](https://docs.anthropic.com/en/docs/test-and-evaluate/strengthen-guardrails/reduce-hallucinations)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Generate an output, then score it

We create an output and apply the checklist to it.

```python
output = ask_claude('Summarize the benefits of remote work in 4 bullets, and include one statistic.')
print(output)
```

## Code-Along: Step 2 — A reusable checklist function

Turn the checklist into something you can run against any output.

```python
def evaluate(output):
    checks = [
      'Answers the actual question asked',
      'All requested parts are present (completeness)',
      'Factual claims are verifiable',
      'Numbers/names/dates/citations double-checked',
      'No internal contradictions',
    ]
    print('OUTPUT UNDER REVIEW:\n', output, '\n')
    print('EVALUATION CHECKLIST (mark Y/N yourself):')
    for c in checks:
        print(f'  [ ] {c}')
evaluate(output)
```
