## Concepts

AI speeds up work, but some outputs must be checked by a qualified human before use. **Always require human review when:**

- The output affects **compliance, legal, financial, medical, or safety** decisions (e.g., a compliance-team summary - Sample Q1).
- It will be **published externally** or sent to a customer/executive.
- It involves **people decisions** (hiring, performance, discipline).
- It contains claims you **cannot verify** or data you cannot confirm.
- The stakes of being wrong are **high**.

Low-risk, easily reversible, internal drafts need lighter checking. Think **risk x impact**: higher either one, more human oversight.

Associates also know when to **escalate** more complex/technical work to Architects and Developers.

This maps to Domain 2: *Determine when human review or additional verification is required*.

```python
# !pip install anthropic   # run once if the SDK is not installed
# %pip install python-dotenv  # only if it isn't installed
import os
from dotenv import find_dotenv, load_dotenv

env_path = find_dotenv(".env", usecwd=True)

if not env_path:
    raise FileNotFoundError("Could not find .env from the notebook's working directory")

load_dotenv(env_path)
print("API key loaded:", bool(os.getenv("ANTHROPIC_API_KEY")))


def ask_claude(prompt, system=None, model="claude-sonnet-5-5", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        # Newer models may return thinking blocks first; keep only the text blocks
        return "".join(b.text for b in resp.content if b.type == "text")
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — A simple risk-based decision helper

Answer a few questions to get a recommendation.

```python
def needs_human_review(external, regulated, people_decision, unverifiable, high_stakes):
    reasons = []
    if external: reasons.append('goes to an external/senior audience')
    if regulated: reasons.append('touches compliance/legal/financial/medical/safety')
    if people_decision: reasons.append('affects people decisions')
    if unverifiable: reasons.append('contains unverifiable claims')
    if high_stakes: reasons.append('high cost if wrong')
    if reasons:
        return 'HUMAN REVIEW REQUIRED because: ' + '; '.join(reasons)
    return 'Lighter check OK (low-risk, reversible, internal).'

print(needs_human_review(external=True, regulated=True, people_decision=False, unverifiable=True, high_stakes=True))
print(needs_human_review(external=False, regulated=False, people_decision=False, unverifiable=False, high_stakes=False))
```

```python
always_review = [
  'Anything sent to compliance, legal, or finance',
  'Anything published externally or to executives',
  'Any people decision (hiring, performance)',
  'Any output with claims I cannot verify',
  'Any command that will be run against production workloads'
]
lighter = [
  'Internal brainstorming lists',
  'Rough first drafts I will edit heavily',
  'Confluence document drafts',
]
lines = ['# My human-review decision rule','','## Always require human review'] + [f'- {x}' for x in always_review]
lines += ['','## Lighter check is OK'] + [f'- {x}' for x in lighter]
with open('human_review_rule.md','w',encoding='utf-8') as f:
    f.write('\n'.join(lines))
print('Saved human_review_rule.md')
print('\n'.join(lines))
```

## Key Takeaways

- High risk x high impact = mandatory human review
- Compliance, legal, financial, external, and people decisions always need review
- Unverifiable claims should never go out unchecked (Sample Q1)
- Match the depth of review to the stakes
