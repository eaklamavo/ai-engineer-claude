## Concepts

Beyond outright false facts, two subtler problems matter:

- **Inconsistencies** — the output contradicts itself (a number in paragraph 1 differs from the table; a recommendation conflicts with the stated goal).
- **Bias** — the output unfairly favors or excludes groups, assumes a stereotype, presents one-sided views, or uses non-inclusive language. Bias can come from how a prompt is phrased or from the training data.

Review habits:
- Cross-check facts/numbers that appear in more than one place.
- Ask: *Who is represented? Who is left out? Is the framing one-sided?*
- For decisions affecting people, require a human review (next lesson).

This maps to Domain 2: *Identify hallucinations, inconsistencies, and biases*.

> **Reference:** [Anthropic Docs - Reduce hallucinations & guardrails](https://docs.anthropic.com/en/docs/test-and-evaluate/strengthen-guardrails/reduce-hallucinations)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Check for internal consistency

Ask for content that contains a number in two places, then verify they match.

```python
text = ask_claude('In 3 sentences, describe a team of 12 people, and in a final line give the headcount as a number.')
print(text)
print('\nREVIEW: Does the headcount number match the description? Cross-check any figure that appears twice.')
```

## Code-Along: Step 2 — Ask for a balanced, inclusive version

You can prompt for balance - and still review it yourself.

```python
balanced = ask_claude('Give a balanced 4-bullet pros/cons view of a 4-day work week. Use inclusive, neutral language.')
print(balanced)
print('\nREVIEW: Is any group or viewpoint missing or unfairly characterized?')
```
