## Concepts

Claude is an AI assistant made by Anthropic. For an Associate, think of Claude as a capable **thinking and drafting partner** for business tasks: summarizing documents, drafting emails, analyzing text, brainstorming, and planning.

**Three surfaces you will hear about on the exam:**

- **Chat** — a back-and-forth conversation. Good for quick, one-off tasks.
- **Projects** — a reusable workspace where you add custom instructions and knowledge (files/documents) so Claude has consistent context every time. Great for recurring workflows.
- **Artifacts** — a side panel where Claude produces a self-contained deliverable (a document, table, draft, or piece of code) that you can view, edit, and iterate on separately from the chat.

**What Associates do vs. escalate:** Associates apply Claude to *business and productivity tasks* with minimal guidance. Designing API integrations, agentic systems, or enterprise AI architecture is **out of scope** — that work goes to Claude Architects and Developers.

> **Reference:** [Anthropic Docs - Welcome / Overview](https://docs.anthropic.com)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Ask Claude to explain itself (in plain language)

We'll send a simple prompt. If you don't have an API key set, you'll see a friendly mock response - that's fine for learning.

```python
answer = ask_claude('In 3 bullet points, explain what you (Claude) are good at for a marketing manager. Keep it non-technical.')
print(answer)
```

## Code-Along: Step 2 — Match the task to the right surface

A quick mental model you can reuse. Run this to print a cheat-sheet.

```python
surfaces = {
  'One-off question or quick draft': 'Chat',
  'A recurring workflow needing the same context each time': 'Project (with instructions + knowledge)',
  'A deliverable you want to view/edit on its own': 'Artifact',
}
for task, surface in surfaces.items():
    print(f'- {task}\n    -> Use: {surface}\n')
```
