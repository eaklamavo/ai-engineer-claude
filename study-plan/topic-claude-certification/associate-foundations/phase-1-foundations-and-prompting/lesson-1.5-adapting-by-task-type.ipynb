## Concepts

Different tasks call for different prompting strategies:

- **Analysis** — ask for structured reasoning and evidence; request a table or clear criteria. Give Claude the source data.
- **Research** — ask for organized findings and be explicit that claims must be verifiable; remember Claude can be out of date or wrong, so plan to fact-check (Phase 2).
- **Drafting** — specify audience, tone, length, and format; iterate.
- **Brainstorming** — ask for many diverse options, defer judgment, then narrow down.

Matching strategy to task type is a core Associate skill and maps to Domain 1: *Adapt prompting strategies based on task type*.

> **Reference:** [Anthropic Docs - Prompt engineering overview](https://docs.anthropic.com/en/docs/build-with-claude/prompt-engineering/overview)

```python
# !pip install anthropic   # run once if the SDK is not installed
import os

def ask_claude(prompt, system=None, model="claude-3-5-haiku-latest", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Four task types, four prompts

See how the ask changes with the goal.

```python
prompts = {
 'Analysis': 'Analyze these 3 sales figures (Jan 100, Feb 90, Mar 130). Give the trend and one likely driver in a 2-row table.',
 'Research': 'List 3 common employee-onboarding best practices. Mark any claim I should verify before relying on it.',
 'Drafting': 'Draft a 2-sentence welcome message for new hires. Warm, plain language.',
 'Brainstorming': 'Give me 8 varied names for an internal wellness program. No explanations.',
}
for t, p in prompts.items():
    print(f'=== {t} ===')
    print(ask_claude(p))
    print()
```
