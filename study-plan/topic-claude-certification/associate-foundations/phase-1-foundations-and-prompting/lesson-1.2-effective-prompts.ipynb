## Concepts

A good prompt is clear, specific, and gives Claude what it needs. A reliable structure to remember:

- **Role** — who Claude should act as (e.g., 'You are an operations analyst').
- **Task** — what you want done, stated as a clear instruction.
- **Context** — background, audience, source material, constraints.
- **Format** — how the answer should look (bullets, table, email, word count).

Vague prompts produce vague output. The fastest quality win is adding **context** (who it's for, what matters) and a **format** (exactly how you want it back).

This maps to Domain 1: *Create effective prompts for business and technical tasks*.

> **Reference:** [Anthropic Docs - Prompt engineering overview](https://docs.anthropic.com/en/docs/build-with-claude/prompt-engineering/overview)

```python
import os
from dotenv import find_dotenv, load_dotenv

env_path = find_dotenv(".env", usecwd=True)

if not env_path:
    raise FileNotFoundError("Could not find .env from the notebook's working directory")

load_dotenv(env_path)
print("API key loaded:", bool(os.getenv("ANTHROPIC_API_KEY")))

def ask_claude(prompt, system=None, model="claude-sonnet-5-5", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        # Some models return a thinking block first, so join only the text blocks
        return "".join(b.text for b in resp.content if b.type == "text")
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Compare a weak prompt vs. a structured prompt

Notice how much the structure changes the request.

```python
weak = 'Write about our new product.'
strong = (
  'Role: You are a B2B marketing copywriter.\n'
  'Task: Write a short product announcement.\n'
  'Context: Product = TimeTrack, a time-logging app for small teams. Audience = busy office managers.\n'
  'Format: 3 short paragraphs, friendly tone, end with one call to action.'
)
print('--- WEAK ---'); print(ask_claude(weak))
print('\n--- STRONG ---'); print(ask_claude(strong))
```

## Code-Along: Step 2 — A fill-in-the-blank template

Use this skeleton for almost any task.

```python
def build_prompt(role, task, context, fmt):
    return f'Role: {role}\nTask: {task}\nContext: {context}\nFormat: {fmt}'

p = build_prompt(
  role='an HR communications specialist',
  task='Draft an announcement about a new flexible-hours policy',
  context='Audience = all employees; tone = warm and clear; policy starts next month',
  fmt='Under 150 words, plain language, one bullet list of key points')
print(p)
print('\n--- RESPONSE ---')
print(ask_claude(p))
```
