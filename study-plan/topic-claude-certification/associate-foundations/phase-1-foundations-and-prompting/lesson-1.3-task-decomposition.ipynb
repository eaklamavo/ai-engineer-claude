# Lesson 1.3 — Breaking Big Tasks into Steps (Task Decomposition)

**Level: Beginner**

## Learning Objectives

By the end of this lesson, you will be able to:

- Explain why large, complex requests often produce weaker results
- Decompose a complex task into an ordered sequence of smaller prompts
- Chain outputs so each step feeds the next

**You will produce:** A saved multi-step 'task plan' for a real workflow and its chained outputs

## Concepts

When a request is big and complex, Claude can miss parts of it. **Task decomposition** means breaking one large request into a logical sequence of smaller, focused steps.

Benefits:
- Each step is easier to get right and to check.
- You can review and correct between steps.
- The final result is more complete and accurate.

Example: 'Write a full market report' becomes (1) list the sections, (2) draft each section, (3) write the executive summary last from the drafted sections.

This maps to Domain 1: *Apply task decomposition techniques to structure complex requests*.

> **Reference:** [Anthropic Docs - Chain complex prompts](https://docs.anthropic.com/en/docs/build-with-claude/prompt-engineering/chain-prompts)

```python
import os
from dotenv import find_dotenv, load_dotenv

env_path = find_dotenv(".env", usecwd=True)

if not env_path:
    raise FileNotFoundError("Could not find .env from the notebook's working directory")

load_dotenv(env_path)
print("API key loaded:", bool(os.getenv("ANTHROPIC_API_KEY")))

def ask_claude(prompt, system=None, model="claude-sonnet-5-5", max_tokens=600):
    """Send a prompt to Claude. Falls back to a printed mock if no API key is set,
    so this notebook always runs. Set ANTHROPIC_API_KEY in your environment to use the real API.
    NOTE: Verify current model names at https://docs.anthropic.com (they change over time)."""
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("[No ANTHROPIC_API_KEY found - showing a MOCK response so the lesson still runs.]\n")
        return "[MOCK CLAUDE OUTPUT]\nThis is a placeholder. With a real key, Claude's answer appears here."
    try:
        import anthropic
        client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from the environment
        kwargs = {"model": model, "max_tokens": max_tokens,
                  "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        # Some models return a thinking block first, so join only the text blocks
        return "".join(b.text for b in resp.content if b.type == "text")
    except Exception as e:
        print(f"[Could not reach the API: {e}. Showing a mock response.]\n")
        return "[MOCK CLAUDE OUTPUT] Placeholder response."

print("Helper ready. Call ask_claude('your prompt here').")
```

## Code-Along: Step 1 — Turn one big task into steps

We ask Claude to plan before doing - a useful habit.

```python
big_task = 'Create a customer onboarding guide for our software.'
plan = ask_claude(f'Break this task into 4-6 ordered steps I can tackle one at a time. Just list the steps.\nTask: {big_task}')
print(plan)
```

## Code-Along: Step 2 — Chain two steps together

The output of step 1 becomes the input to step 2.

```python
outline = ask_claude('List 4 section headings for a one-page customer onboarding guide. Headings only.')
print('--- OUTLINE ---'); print(outline)
draft = ask_claude(f'Using these headings, write one short sentence under each. Keep it simple.\n{outline}')
print('\n--- DRAFT ---'); print(draft)
```

## Exercise

Decompose one of your real tasks. Fill in the TODO, generate a step plan, and save it.

```python
my_big_task = 'Test a new AI agent for SRE tasks'
steps = ask_claude(f'Break this into 4 ordered, concrete steps. Numbered list only.\nTask: {my_big_task}')
with open('my_task_plan.md', 'w', encoding='utf-8') as f:
    f.write(f'# Task plan: {my_big_task}\n\n{steps}')
print('Saved my_task_plan.md\n')
print(steps)
```

## Key Takeaways

- Large requests are more reliable when split into ordered steps
- Ask Claude to plan the steps first, then execute them
- Chaining lets each step build on a reviewed previous output
- Decomposition makes it easier to check work along the way
