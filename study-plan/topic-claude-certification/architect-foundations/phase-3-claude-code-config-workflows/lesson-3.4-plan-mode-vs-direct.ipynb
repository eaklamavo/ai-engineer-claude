## Concepts

**Plan mode** is for complex tasks: large-scale changes, multiple valid approaches, architectural decisions, multi-file modifications. It enables safe exploration and design before committing to changes, preventing costly rework. Sample Question 5 (monolith -> microservices across dozens of files) is a textbook plan-mode case.

**Direct execution** is for simple, well-scoped changes - a single-file bug fix with a clear stack trace, or adding one validation conditional.

The **Explore subagent** handles verbose discovery phases, preventing context-window exhaustion during multi-phase tasks. A common pattern: **plan mode to investigate, then direct execution to implement** the planned approach.

## Code-Along: Step 1 — A decision function

Score complexity signals; architectural or multi-approach work goes to plan mode.

```python
def mode_for(task):
    signals = [task.get('many_files'), task.get('multiple_approaches'),
               task.get('architectural')]
    return 'plan' if any(signals) else 'direct'

monolith = {'many_files': True, 'multiple_approaches': True, 'architectural': True}
bugfix = {'many_files': False, 'multiple_approaches': False, 'architectural': False}
print('monolith->microservices:', mode_for(monolith))
print('single-file bug fix:', mode_for(bugfix))
```

## Code-Along: Step 2 — Explore subagent for discovery

Delegate verbose discovery so the main session's context stays lean.

```python
def plan_with_explore(task):
    steps = []
    if task.get('verbose_discovery'):
        steps.append('spawn Explore subagent for codebase discovery')
    steps.append('plan mode: design approach')
    steps.append('direct execution: implement planned changes')
    return steps

for s in plan_with_explore({'verbose_discovery': True}):
    print('-', s)
```

## Code-Along: Step 3 — Validate against sample scenarios

We encode the exam's examples as test cases to lock in the judgment.

```python
cases = {
    'library migration (45+ files)': {'many_files': True},
    'add date validation conditional': {},
    'choose integration approach': {'multiple_approaches': True},
}
for name, t in cases.items():
    print(f"{name}: {mode_for(t)}")
```
