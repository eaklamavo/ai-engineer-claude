## Concepts

The **CLAUDE.md hierarchy**:
- **User-level** `~/.claude/CLAUDE.md` - applies only to that user; **not** shared via VCS.
- **Project-level** `.claude/CLAUDE.md` or root `CLAUDE.md` - shared with the team.
- **Directory-level** subdirectory `CLAUDE.md` files - scoped to that directory.

Sample Question-style pitfall: a new teammate doesn't get instructions because they live in **user-level** config rather than project-level. Put shared standards at the project level.

- **`@import`** references external files to keep CLAUDE.md modular (import only the standards relevant to each package).
- **`.claude/rules/`** organizes topic-specific rule files (testing.md, api-conventions.md, deployment.md) instead of one monolith.
- Use **`/memory`** to verify which memory files are loaded when behavior looks inconsistent.

> **Reference:** [Anthropic - Claude Code memory/CLAUDE.md](https://docs.anthropic.com/en/docs/claude-code/memory)

```python
import os, pathlib
ROOT = pathlib.Path('cc_project'); ROOT.mkdir(exist_ok=True)
os.chdir(ROOT)
(pathlib.Path('.claude') / 'rules').mkdir(parents=True, exist_ok=True)
(pathlib.Path('.claude') / 'commands').mkdir(parents=True, exist_ok=True)
(pathlib.Path('.claude') / 'skills').mkdir(parents=True, exist_ok=True)
print('project root:', os.getcwd())
```

## Code-Along: Step 1 — Write a modular project CLAUDE.md with @import

Shared standards live at project level; @import pulls in focused files.

```python
claude_md = '''# Project: Orders Service

## Universal Standards
- Python 3.12, type hints required
- All public functions need docstrings

## Imported Standards
@import .claude/rules/testing.md
@import .claude/rules/api-conventions.md
'''
pathlib.Path('CLAUDE.md').write_text(claude_md, encoding='utf-8')
print(pathlib.Path('CLAUDE.md').read_text(encoding='utf-8'))
```

## Code-Along: Step 2 — Split topics into .claude/rules/

Each concern gets its own file - easier to maintain than one giant CLAUDE.md.

```python
rules = {
    'testing.md': '# Testing\n- Use pytest\n- One assert-focused test per behavior\n',
    'api-conventions.md': '# API\n- REST, snake_case JSON\n- Return RFC-7807 errors\n',
    'deployment.md': '# Deployment\n- Blue/green only\n- Tag releases vMAJOR.MINOR\n',
}
for name, body in rules.items():
    (pathlib.Path('.claude/rules') / name).write_text(body, encoding='utf-8')
print([p.name for p in pathlib.Path('.claude/rules').iterdir()])
```

## Code-Along: Step 3 — Diagnose a hierarchy problem

Instructions a teammate can't see are usually in user scope. We classify by location.

```python
def scope_of(path):
    if path.startswith('~/.claude') or path.startswith(os.path.expanduser('~')):
        return 'user (NOT shared via VCS)'
    return 'project (shared via VCS)'

print('~/.claude/CLAUDE.md ->', scope_of('~/.claude/CLAUDE.md'))
print('CLAUDE.md ->', scope_of('CLAUDE.md'))
print('Fix: move shared team standards to project-level CLAUDE.md.')
```
