## Concepts

Techniques for progressive improvement:

- **Concrete input/output examples** communicate expected transformations far better than prose when natural-language descriptions are interpreted inconsistently (2-3 examples).
- **Test-driven iteration:** write a test suite first (behavior, edge cases, performance), then iterate by sharing **test failures** to guide the model.
- **Interview pattern:** have Claude ask questions to surface considerations you may not have anticipated (cache invalidation strategies, failure modes) **before** implementing.
- **Batching fixes:** provide all issues in a single message when they **interact**; fix **independent** problems sequentially.

## Code-Along: Step 1 — Communicate with I/O examples

Concrete pairs pin down an ambiguous transformation.

```python
examples = [
    {"input": "2026-11-11", "output": "Nov 11, 2026"},
    {"input": "", "output": "(no date)"},
    {"input": "not-a-date", "output": "(invalid date)"},
]
def examples_block(exs):
    return "\n".join(f"in={e['input']!r} -> out={e['output']!r}" for e in exs)
print(examples_block(examples))
```

## Code-Along: Step 2 — TDD failure-feedback loop

Share failing tests, not prose, to steer the next iteration.

```python
def run_tests(fmt):
    failures = []
    if fmt('') != '(no date)':
        failures.append('empty string should render (no date)')
    if fmt('not-a-date') != '(invalid date)':
        failures.append('invalid input should render (invalid date)')
    return failures

naive = lambda s: s  # first attempt
print('iteration 1 failures:', run_tests(naive))
# Feed these failures back as the refinement prompt for the next iteration.
```

## Code-Along: Step 3 — Interview pattern + fix batching

Ask clarifying questions first; then batch interacting fixes, sequence independent ones.

```python
interview_prompt = (
    "Before implementing the cache layer, ask me about: invalidation strategy, "
    "eviction policy, failure modes, and consistency requirements."
)
def fix_strategy(issues_interact):
    return 'single detailed message' if issues_interact else 'sequential iteration'

print(interview_prompt)
print('interacting issues ->', fix_strategy(True))
print('independent issues ->', fix_strategy(False))
```
